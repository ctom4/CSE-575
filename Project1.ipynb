```python
# Load the generated training and test sets for the project calculations.
myID = '4853'

train0 = scipy.io.loadmat(
    'digit0_stu_train' + myID + '.mat'
)['target_img']

train1 = scipy.io.loadmat(
    'digit1_stu_train' + myID + '.mat'
)['target_img']

test0 = scipy.io.loadmat(
    'digit0_testset.mat'
)['target_img']

test1 = scipy.io.loadmat(
    'digit1_testset.mat'
)['target_img']
```

```python
import numpy as np

# Task 1: Extract two features from each image.
# Each image has shape 28 x 28. Axes 1 and 2 contain the pixel values.
def extract_feature1(images):
    # Feature 1: average pixel brightness for each image.
    return np.mean(images, axis=(1, 2))

def extract_feature2(images):
    # Feature 2: standard deviation of pixel brightness for each image.
    return np.std(images, axis=(1, 2))

# Extract the same two features from both training classes.
train0_feature1 = extract_feature1(train0)
train0_feature2 = extract_feature2(train0)
train1_feature1 = extract_feature1(train1)
train1_feature2 = extract_feature2(train1)

# The test images must be converted to the same two-feature representation
# before they can be classified in Task 3.
test0_feature1 = extract_feature1(test0)
test0_feature2 = extract_feature2(test0)
test1_feature1 = extract_feature1(test1)
test1_feature2 = extract_feature2(test1)
```

```python
# Task 2: Estimate the Gaussian parameters using only the training data.
# Each class has one Gaussian distribution for Feature 1 and another for
# Feature 2. Each Gaussian requires a mean and a variance.

# Digit 0, Feature 1
mean_feature1_digit0 = np.mean(train0_feature1)
var_feature1_digit0 = np.var(train0_feature1)

# Digit 0, Feature 2
mean_feature2_digit0 = np.mean(train0_feature2)
var_feature2_digit0 = np.var(train0_feature2)

# Digit 1, Feature 1
mean_feature1_digit1 = np.mean(train1_feature1)
var_feature1_digit1 = np.var(train1_feature1)

# Digit 1, Feature 2
mean_feature2_digit1 = np.mean(train1_feature2)
var_feature2_digit1 = np.var(train1_feature2)
```

```python
# Task 3: Implement the Gaussian Naive Bayes classifier.

# Gaussian probability density function using the mean and variance
# estimated from the training data in Task 2.
def gaussian_pdf(x, mean, variance):
    return (
        1.0 / np.sqrt(2.0 * np.pi * variance)
    ) * np.exp(
        -((x - mean) ** 2) / (2.0 * variance)
    )


# The project specifies equal prior probabilities for digits 0 and 1.
prior_digit0 = 0.5
prior_digit1 = 0.5


# Predict class labels using the Gaussian Naive Bayes scores.
def predict_digit(feature1, feature2):

    # Naive Bayes score for digit 0:
    # P(y=0) * p(feature1 | y=0) * p(feature2 | y=0)
    score_digit0 = (
        prior_digit0
        * gaussian_pdf(
            feature1,
            mean_feature1_digit0,
            var_feature1_digit0
        )
        * gaussian_pdf(
            feature2,
            mean_feature2_digit0,
            var_feature2_digit0
        )
    )

    # Naive Bayes score for digit 1:
    # P(y=1) * p(feature1 | y=1) * p(feature2 | y=1)
    score_digit1 = (
        prior_digit1
        * gaussian_pdf(
            feature1,
            mean_feature1_digit1,
            var_feature1_digit1
        )
        * gaussian_pdf(
            feature2,
            mean_feature2_digit1,
            var_feature2_digit1
        )
    )

    # Predict the class with the larger Naive Bayes score.
    return np.where(score_digit0 > score_digit1, 0, 1)


# Predict labels for all digit-0 and digit-1 test images.
predicted_digit0 = predict_digit(
    test0_feature1,
    test0_feature2
)

predicted_digit1 = predict_digit(
    test1_feature1,
    test1_feature2
)
```

```python
# Task 4: Calculate class-specific test accuracy.

# Every image in test0 is a digit 0, so a prediction of 0 is correct.
accuracy_digit0 = np.mean(predicted_digit0 == 0)

# Every image in test1 is a digit 1, so a prediction of 1 is correct.
accuracy_digit1 = np.mean(predicted_digit1 == 1)
```
